```python
df['Genre'].unique()
```

```python
cat_counts=df['Genre'].value_counts()
cat_counts
```

```python
x_bar = cat_counts.index
x_bar
```

```python
y_bar = cat_counts.array
y_bar
```

```python
plt.bar(x_bar, y_bar)
```

```python
plt.figure(figsize=(12,8))
plt.bar(x_bar, y_bar)
```

```python
plt.figure(figsize=(12,8))
plt.bar(x_bar, y_bar, width=0.5, color='green')
plt.xticks(rotation=60, fontsize=10)
plt.xlabel('Genre', fontsize=10)
plt.ylabel('Counts', fontsize=10)
plt.title('Top Genre Analysis', fontsize=18)
plt.show()
```
