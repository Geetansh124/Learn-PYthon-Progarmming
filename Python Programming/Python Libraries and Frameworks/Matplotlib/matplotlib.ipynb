```python
#plt.plot(x,y color='color name', lifestyle='line_Style', linewidth=value, marker='marker symbol', lable = label name')
#plt.xlabel('text')

import matplotlib.pyplot as plt


months = [1, 2, 3, 4]
sales = [1000, 1500, 1200, 1800]


plt.plot(months, sales, color='red', linestyle='--', marker='o', label='2025 sales data')


plt.legend()


plt.xlabel('Months')
plt.ylabel('Sales')
plt.title('Monthly Sales Performance')
plt.legend(loc='upper left', fontsize=12)
plt.grid(color='lightblue', linestyle=':', linewidth=0.5)
plt.xlim(1,4)
plt.ylim(0, 2000)
plt.xticks([1,2,3,4], ['M1', 'M2', 'M3', 'M4'])
plt.show()
```

```python
#plt.pie(values, labels= labels_list, coloer= color_list, auto='%1.1%%')

import matplotlib.pyplot as plt

regions = ('North', 'South', 'East', 'West' )
revenue = [3000, 2000, 1500, 1000]

plt.pie(revenue, labels=regions, autopct='%1.1f%%', colors=['blue', 'green', 'yellow', 'red', 'magenta' ])
plt.title('Revenue Contribuion By Region')
plt.show()
```

*Sub Plot*

```python
#plt.subplot(nrows, ncols, index)

import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,20,15,25]

plt.subplot(1,2,1) #1row, 2column, 1nd suplot
plt.plot(x,y)
plt.title('Line Chart')

plt.subplot(1,2,2) #1row, 2column, 2nd suplot
plt.bar(x,y)
plt.title('Bar Chart')

plt.show()
```

```python
import matplotlib.pyplot as plt

#fig, ax = plt.subplots(nrows, ncols, figsize=(width, height)
fig, ax = plt.subplots(1,2, figsize = (10,5))

x = [1,2,3,4]
y = [10,20,15,25]

ax[0].plot(x,y, color='Coral')
ax[0].set_title('Line Chart')

ax[1].bar(x,y, color = 'Teal')
ax[1].set_title('Bar Chart')

fig.suptitle('Comparision of Line and Bar Chart')
plt.show()

plt.tight_layout()
plt.savefig('E:\\Phone Link\\Line-Bar_Chart.pdf', dpi = 300, bbox_inches = 'tight')
plt.show()
```
